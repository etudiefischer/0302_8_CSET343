```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    roc_curve, roc_auc_score
)

sns.set(style="whitegrid")
%matplotlib inline
```

```python
# Original UCI Breast Cancer Wisconsin (Diagnostic) dataset
col_names = ['id', 'diagnosis',
    'radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean', 'smoothness_mean',
    'compactness_mean', 'concavity_mean', 'concave_points_mean', 'symmetry_mean', 'fractal_dimension_mean',
    'radius_se', 'texture_se', 'perimeter_se', 'area_se', 'smoothness_se',
    'compactness_se', 'concavity_se', 'concave_points_se', 'symmetry_se', 'fractal_dimension_se',
    'radius_worst', 'texture_worst', 'perimeter_worst', 'area_worst', 'smoothness_worst',
    'compactness_worst', 'concavity_worst', 'concave_points_worst', 'symmetry_worst', 'fractal_dimension_worst']

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/wdbc.data"
df = pd.read_csv(url, header=None, names=col_names)
df = df.drop(columns='id')

print("Shape:", df.shape)
df.head()
```

```python
df.describe().T
```

```python
# Missing values
print("Missing values per column:\n", df.isnull().sum())

# Duplicates
dup_count = df.duplicated().sum()
print(f"\nDuplicate rows: {dup_count}")
df = df.drop_duplicates()

# Outlier detection using IQR method
def count_outliers_iqr(series):
    Q1, Q3 = series.quantile(0.25), series.quantile(0.75)
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
    return ((series < lower) | (series > upper)).sum()

numeric_cols = df.select_dtypes(include=np.number).columns
outlier_summary = df[numeric_cols].apply(count_outliers_iqr).sort_values(ascending=False)
print("\nTop 10 features by outlier count (IQR method):")
print(outlier_summary.head(10))

# Moderate outliers are kept — medical measurements can legitimately be extreme
# (e.g., large tumors), so we flag them but don't remove them.
```

```python
key_features = ['radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean',
                 'smoothness_mean', 'concavity_mean']

df[key_features].hist(bins=20, figsize=(14, 8))
plt.suptitle("Histograms of Key Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))
sns.boxplot(data=df[key_features])
plt.xticks(rotation=30)
plt.title("Box Plots of Key Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18, 14))
corr = df[numeric_cols].corr()
sns.heatmap(corr, cmap='coolwarm', annot=False, center=0)
plt.title("Correlation Heatmap of All Features")
plt.show()
```

```python
plt.figure(figsize=(5, 4))
sns.countplot(x='diagnosis', data=df)
plt.title("Class Distribution")
plt.xlabel("Diagnosis (M = Malignant, B = Benign)")
plt.ylabel("Count")
plt.show()

print(df['diagnosis'].value_counts(normalize=True) * 100)
```

**Discussion — Class Imbalance:**
The dataset is moderately imbalanced (roughly 63% benign vs. 37% malignant). This can
bias a model toward the majority class, inflating accuracy while under-detecting
malignant cases — the costlier error in a medical context. This motivates:
- Using stratified sampling in the train/test split.
- Looking beyond accuracy to precision, recall, and F1-score per class, with special
  attention to recall on the malignant class (minimizing false negatives).

```python
# diagnosis: M = Malignant, B = Benign -> encode M=1, B=0
le = LabelEncoder()
df['Outcome'] = le.fit_transform(df['diagnosis'])   # B -> 0, M -> 1
print(dict(zip(le.classes_, le.transform(le.classes_))))

X = df.drop(columns=['diagnosis', 'Outcome'])
y = df['Outcome']

print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

print("Train shape:", X_train.shape, " Test shape:", X_test.shape)
print("\nTrain class balance:\n", y_train.value_counts(normalize=True))
print("\nTest class balance:\n", y_test.value_counts(normalize=True))
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)

X_train_scaled.describe().T[['mean', 'std']].head()
```

```python
log_reg = LogisticRegression(max_iter=5000, random_state=42)
log_reg.fit(X_train_scaled, y_train)

y_pred = log_reg.predict(X_test_scaled)
y_proba = log_reg.predict_proba(X_test_scaled)[:, 1]
```

```python
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc:.4f}")
```

```python
print(classification_report(y_test, y_pred, target_names=['Benign (0)', 'Malignant (1)']))
```

```python
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Benign', 'Malignant'])
disp.plot(cmap='Blues', values_format='d')
plt.title("Confusion Matrix")
plt.show()
```

```python
auc = roc_auc_score(y_test, y_proba)
fpr, tpr, thresholds = roc_curve(y_test, y_proba)

plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f"Logistic Regression (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label="Random guess")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()

print(f"AUC Score: {auc:.4f}")
```

**Interpretation of Results:**

- **Accuracy** shows overall correctness but can mask imbalance issues.
- **Precision (malignant)**: of predicted malignant cases, how many were correct —
  low precision means more false alarms.
- **Recall (malignant)**: of actual malignant cases, how many were caught — the most
  critical metric here, since a missed cancer (false negative) is far costlier than
  a false alarm.
- **F1-score** balances precision and recall per class.
- **Confusion matrix** shows exactly where errors occur (false positives vs. false
  negatives), which matters more clinically than the single accuracy number.
- **ROC/AUC** measures separability across all thresholds; a high AUC (typically
  >0.95 on this dataset) suggests the decision threshold could even be tuned lower
  than 0.5 to further favor recall on malignant cases.
